```python
X_min = X.min(axis=0)
X_max = X.max(axis=0)
```

```python
X_normal = (X-X_min)/(X_max-X_min)  
```

```python
X_mu=X.mean(axis=0) 
X_sigma =X.std(axis=0) 
X_standart = (X - X_mu) / X_sigma
```

```python
print(np.round(X_standart.mean(axis=0))) 
print(X_standart.std(axis=0))
```

```python
kovaryans_matrisi = np.cov(X_standart, rowvar=False)
print(kovaryans_matrisi)
```

```python
korelasyon_matrisi = np.corrcoef(X_standart, rowvar=False)
print(korelasyon_matrisi)
```

```python
kalite_korelasyonlari = np.corrcoef(X_standart.T, y)
puanlar = kalite_korelasyonlari[-1, :-1]
en_guclu_3_indeks = np.argsort(np.abs(puanlar))[-3:]
print("En yüksek korelasyona sahip 3 sütunun indeksleri:", en_guclu_3_indeks)
```

```python
X_en_iyi = X_standart[:, en_guclu_3_indeks]
```

```python
print(X_en_iyi.shape)
```

```python
satir_sayisi = X_en_iyi.shape[0]  
birler_sutunu = np.ones((satir_sayisi, 1))  
X_bias = np.hstack((birler_sutunu, X_en_iyi))
theta = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y
print("Modelin bulduğu altın ağırlıklar (Theta):")
print(theta)
```

```python
y_pred = X_bias @ theta
mse = np.mean((y - y_pred) ** 2)
r2 = 1 - (np.sum((y - y_pred) ** 2) / np.sum((y - y.mean()) ** 2))
print("Ortalama Kare Hata (MSE):", mse)
print("Modelin Başarı Skoru (R^2):", r2)
```
